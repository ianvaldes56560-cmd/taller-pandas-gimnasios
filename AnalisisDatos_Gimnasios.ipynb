```python
import pandas as pd
import numpy as np
```

## 1. Lectura y verificación de carga

```python
df_gimnasio = pd.read_csv("gimnasios_entrenamiento.csv")
print(df_gimnasio.shape)
print(list(df_gimnasio.columns))
df_gimnasio.head()
```

```python
df_gimnasio.info()
```

## 2. Selección unidimensional y multivariable

```python
disciplinas = df_gimnasio["Disciplina_Principal"]
print(type(disciplinas))
disciplinas.head()
```

```python
df_reducido = df_gimnasio[["Sede", "Plan", "Visitas_Mes"]]
print(type(df_reducido))
df_reducido.head()
```

## 3. Acceso por etiqueta con .loc[]

```python
df_gimnasio.loc[350]
```

## 4. Tratamiento de redundancias

```python
print(df_gimnasio.duplicated().sum())
print(len(df_gimnasio))
df_gimnasio = df_gimnasio.drop_duplicates(keep="first")
print(df_gimnasio.duplicated().sum())
print(len(df_gimnasio))
```

## 5. Filtrado por nivel de actividad

```python
alta_frecuencia = df_gimnasio[df_gimnasio["Visitas_Mes"] > 18]
print(len(alta_frecuencia))
alta_frecuencia.head()
```

## 6. Filtrado categórico con .isin()

6


```python
planes_premium = df_gimnasio[df_gimnasio["Plan"].isin(["VIP Total", "Platino"])]
print(planes_premium["Plan"].value_counts())
planes_premium.head()
```

## 7. Filtrado por intervalo con .between()

```python
rango_duracion = df_gimnasio[df_gimnasio["Minutos_Promedio_Sesion"].between(45.0, 75.0)]
print(len(rango_duracion))
rango_duracion.head()
```

## 8. Consultas directas con .query()

```python
viterbo_activos = df_gimnasio.query('Sede == "Viterbo Centro" and Estado_Membresia == "Activa"')
print(len(viterbo_activos))
viterbo_activos.head()
```

## 9. Filtrado compuesto con métodos de texto

```python
cond_sede = df_gimnasio["Sede"].str.startswith("Anserma")
cond_gasto = df_gimnasio["Gasto_Suplementos"] > 50000
anserma_gasto_alto = df_gimnasio[cond_sede & cond_gasto]
print(len(anserma_gasto_alto))
anserma_gasto_alto.head()
```

## 10. Imputación de faltantes con .fillna()

```python
print(df_gimnasio[["Minutos_Promedio_Sesion", "Gasto_Suplementos"]].isna().sum())
promedio_minutos = df_gimnasio["Minutos_Promedio_Sesion"].mean()
print(round(promedio_minutos, 2))
df_gimnasio["Minutos_Promedio_Sesion"] = df_gimnasio["Minutos_Promedio_Sesion"].fillna(promedio_minutos)
df_gimnasio["Gasto_Suplementos"] = df_gimnasio["Gasto_Suplementos"].fillna(0.0)
print(df_gimnasio[["Minutos_Promedio_Sesion", "Gasto_Suplementos"]].isna().sum())
```

## 11. Transformación condicional con .apply()

```python
def clasificar_perfil(visitas):
    if visitas < 8:
        return "Esporádico"
    elif visitas <= 16:
        return "Regular"
    else:
        return "Frecuente"

df_gimnasio["Perfil_Atleta"] = df_gimnasio["Visitas_Mes"].apply(clasificar_perfil)
df_gimnasio[["Visitas_Mes", "Perfil_Atleta"]].head(10)
```

```python
df_gimnasio["Perfil_Atleta"].value_counts()
```

## 12. Renombrado y reordenamiento de atributos

```python
df_gimnasio = df_gimnasio.rename(columns={
    "Minutos_Promedio_Sesion": "Duracion_Media",
    "Disciplina_Principal": "Deporte",
})
prioritarias = ["ID_Socio", "Perfil_Atleta"]
resto = [c for c in df_gimnasio.columns if c not in prioritarias]
df_gimnasio = df_gimnasio[prioritarias + resto]
df_gimnasio.head()
```

## 13. Agregación por centro de entrenamiento

```python
resumen_sede = df_gimnasio.groupby("Sede").agg(
    Duracion_Media_Promedio=("Duracion_Media", "mean"),
    Gasto_Total=("Gasto_Suplementos", "sum"),
)
resumen_sede
```

## 14. Métricas de dispersión personalizadas

```python
def rango(serie):
    return serie.max() - serie.min()

rango_por_plan = df_gimnasio.groupby("Plan")[["Visitas_Mes", "Duracion_Media"]].agg(rango)
rango_por_plan
```

## 15. Agrupamiento multinivel

```python
multinivel = df_gimnasio.groupby(["Sede", "Plan"]).agg(
    Visitas_Promedio=("Visitas_Mes", "mean"),
    Gasto_Total=("Gasto_Suplementos", "sum"),
)
multinivel
```

```python
import pandas as pd
import numpy as np
```

## Creación de los datos

```python
np.random.seed(2026)
n = 985

data = {
    "ID_Socio": [f"SOC-{10001 + i}" for i in range(n)],
    "Sede": np.random.choice(["Manizales Centro", "Pereira Circunvalar", "Armenia Norte", "Chinchiná", "Medellín Poblado"], n, p=[0.28, 0.22, 0.18, 0.12, 0.20]),
    "Plan": np.random.choice(["Básico", "Platino", "VIP Total", "Estudiantil"], n, p=[0.35, 0.25, 0.15, 0.25]),
    "Disciplina_Principal": np.random.choice(["Musculación", "Crossfit", "Spinning", "Yoga y Pilates", "Funcional"], n),
    "Visitas_Mes": np.random.randint(1, 29, n),
    "Minutos_Promedio_Sesion": np.random.normal(60, 18, n).clip(20, 120).round(1),
    "Gasto_Suplementos": np.random.choice([0, 15000, 20000, 25000, 35000, 45000, 60000, 75000, 90000, 120000], n).astype(float),
    "Estado_Membresia": np.random.choice(["Activa", "Congelada", "Vencida"], n, p=[0.70, 0.15, 0.15]),
}

df_base = pd.DataFrame(data)
df_base
```

```python
df_base.loc[np.random.choice(n, 60, replace=False), "Minutos_Promedio_Sesion"] = np.nan
df_base.loc[np.random.choice(n, 90, replace=False), "Gasto_Suplementos"] = np.nan
df_base.isna().sum()
```

```python
df_completo = pd.concat([df_base, df_base.sample(15, random_state=7)], ignore_index=True)
df_completo = df_completo.sample(frac=1, random_state=11).reset_index(drop=True)
df_completo.shape
```

```python
df_completo.to_csv("gimnasios_entrenamiento.csv", index=False)
```

## 1. Lectura y verificación de carga

```python
df_gimnasio = pd.read_csv("gimnasios_entrenamiento.csv")
df_gimnasio
```

```python
df_gimnasio.info()
```

## 2. Selección unidimensional y multivariable

```python
disciplinas = df_gimnasio["Disciplina_Principal"]
disciplinas, type(disciplinas)
```

```python
df_reducido = df_gimnasio[["Sede", "Plan", "Visitas_Mes"]]
df_reducido
```

## 3. Acceso por etiqueta con .loc[]

```python
fila = df_gimnasio.loc[350]
fila
```

## 4. Tratamiento de redundancias

```python
df_gimnasio.duplicated().sum()
```

```python
df_sin_duplicados = df_gimnasio.drop_duplicates(keep="first")
df_sin_duplicados.shape
```

## 5. Filtrado por nivel de actividad

```python
alta_frecuencia = df_sin_duplicados[df_sin_duplicados["Visitas_Mes"] > 18]
alta_frecuencia
```

## 6. Filtrado categórico con .isin()

```python
planes_premium = df_sin_duplicados[df_sin_duplicados["Plan"].isin(["VIP Total", "Platino"])]
planes_premium
```

## 7. Filtrado por intervalo con .between()

```python
rango_duracion = df_sin_duplicados[df_sin_duplicados["Minutos_Promedio_Sesion"].between(45.0, 75.0)]
rango_duracion
```

## 8. Consultas directas con .query()

```python
medellin_activos = df_sin_duplicados.query('Sede == "Medellín Poblado" and Estado_Membresia == "Activa"')
medellin_activos
```

## 9. Filtrado compuesto con métodos de texto

```python
filtro = (df_sin_duplicados["Sede"].str.startswith("Manizales")) & (df_sin_duplicados["Gasto_Suplementos"] > 50000)
manizales_gasto_alto = df_sin_duplicados[filtro]
manizales_gasto_alto
```

## 10. Imputación de faltantes con .fillna()

```python
df_fill = df_sin_duplicados.copy()
promedio_minutos = df_fill["Minutos_Promedio_Sesion"].mean()
df_fill["Minutos_Promedio_Sesion"] = df_fill["Minutos_Promedio_Sesion"].fillna(promedio_minutos)
df_fill["Gasto_Suplementos"] = df_fill["Gasto_Suplementos"].fillna(0.0)
df_fill.isna().sum()
```

## 11. Transformación condicional con .apply()

```python
def clasificar_perfil(visitas):
    if visitas < 8:
        return "Esporádico"
    elif visitas <= 16:
        return "Regular"
    else:
        return "Frecuente"

df_fill["Perfil_Atleta"] = df_fill["Visitas_Mes"].apply(clasificar_perfil)
df_fill[["Visitas_Mes", "Perfil_Atleta"]]
```

## 12. Renombrado y reordenamiento de atributos

```python
df_renombrado = df_fill.rename(columns={
    "Minutos_Promedio_Sesion": "Duracion_Media",
    "Disciplina_Principal": "Deporte"
})
inicio = ["ID_Socio", "Perfil_Atleta"]
resto = [c for c in df_renombrado.columns if c not in inicio]
df_renombrado = df_renombrado[inicio + resto]
df_renombrado
```

## 13. Agregación por centro de entrenamiento

```python
grouped = df_renombrado.groupby("Sede")
resumen_sede = grouped.agg({
    "Duracion_Media": "mean",
    "Gasto_Suplementos": "sum"
})
resumen_sede
```

## 14. Métricas de dispersión personalizadas

```python
def rango(serie):
    return serie.max() - serie.min()

rango_por_plan = df_renombrado.groupby("Plan")[["Visitas_Mes", "Duracion_Media"]].agg(rango)
rango_por_plan
```

## 15. Agrupamiento multinivel

```python
multinivel = df_renombrado.groupby(["Sede", "Plan"]).agg({
    "Visitas_Mes": "mean",
    "Gasto_Suplementos": "sum"
})
multinivel
```
